### 📊 Notebook Overview
This notebook serves as the **foundational layer** for the Lean Logistics project. Its primary goal is to provision the logical three-level namespace infrastructure and apply initial governance metadata.

### 🎯 Objectives
- **Namespace Provisioning:** Configure and describe the Bronze, Silver, and Gold catalogs.
- **Schema Management:** Create and document domain-specific schemas (`db_logistics`) across the medallion layers.
- **Governance Injection:** Apply Discovery Tags and Business Glossary comments to catalogs for high-level data discovery.

### 🛡️ Governance Standards
- **Standard:** Unity Catalog Three-Level Namespace (`catalog.schema.table`).
- **Project Tagging:** All assets are tagged with `project: lean_logistics` and `environment: dev`.
- **Language:** Technical English for global accessibility.

---

```python
%sql
-- 1. Updating Catalogs Metadata
-- Applying tags and high-level descriptions for Unity Catalog governance

ALTER CATALOG cat_tm_services_bronze SET TAGS ('project' = 'lean_logistics', 'environment' = 'dev');
COMMENT ON CATALOG cat_tm_services_bronze IS 'Raw Layer: Original data ingestion from external sources (Kaggle API). Low-latency, high-fidelity storage.';

ALTER CATALOG cat_tm_services_silver SET TAGS ('project' = 'lean_logistics', 'environment' = 'dev');
COMMENT ON CATALOG cat_tm_services_silver IS 'Refined Layer: Cleaned, deduplicated, and standardized entities. The single source of truth.';

ALTER CATALOG cat_tm_services_gold SET TAGS ('project' = 'lean_logistics', 'environment' = 'dev');
COMMENT ON CATALOG cat_tm_services_gold IS 'Business Layer: Optimized OBT (One Big Table) for logistics KPIs and BI consumption.';
```

```python
%sql
-- 2. Updating Schema Metadata
-- Documenting domain-specific schemas for business context and data discovery

COMMENT ON SCHEMA cat_tm_services_bronze.db_logistics IS 'Main schema for raw logistics data (Olist dataset).';
COMMENT ON SCHEMA cat_tm_services_silver.db_logistics IS 'Cleaned logistics entities (orders, products, etc.) with schema enforcement.';
COMMENT ON SCHEMA cat_tm_services_gold.db_logistics IS 'Consolidated logistics metrics and denormalized views for dashboards.';
```
